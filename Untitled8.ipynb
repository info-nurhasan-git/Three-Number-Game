```python
# Identify all columns starting with 'Unnamed' and fill their missing values with 0
unnamed_cols = [col for col in df.columns if "Unnamed" in col]
df[unnamed_cols] = df[unnamed_cols].fillna(0)
display(df.head(5))
```

```python
# Drop all columns that contain 'Unnamed' to keep only the main dataset columns
main_cols = [col for col in df.columns if 'Unnamed' not in col]
df_clean = df[main_cols].copy()

# Display the cleaned DataFrame shape and head
print(f"New shape: {df_clean.shape}")
display(df_clean.head())
```

```python
df_clean['Payment_Method'].unique()
```

```python
df_clean['Country'].nunique()
```

```python
df_clean['Total_Revenue'] = df_clean['Price'] * df_clean['Quantity']
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set stylistic aesthetics for our data analyst visualizations
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Total Revenue by Country
revenue_by_country = df_clean.groupby('Country')['Total_Revenue'].sum().reset_index().sort_values(by='Total_Revenue', ascending=False)
sns.barplot(x='Total_Revenue', y='Country', data=revenue_by_country, ax=axes[0, 0], palette='Blues_r')
axes[0, 0].set_title('Total Revenue by Country', fontsize=14, fontweight='bold')
axes[0, 0].set_xlabel('Revenue ($)')

# 2. Total Quantity Sold by iPhone Model
quantity_by_model = df_clean.groupby('iPhone_Model')['Quantity'].sum().reset_index().sort_values(by='Quantity', ascending=False)
sns.barplot(x='Quantity', y='iPhone_Model', data=quantity_by_model, ax=axes[0, 1], palette='Greens_r')
axes[0, 1].set_title('Total Units Sold by iPhone Model', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('Units Sold')

# 3. Revenue Share by Payment Method
revenue_by_payment = df_clean.groupby('Payment_Method')['Total_Revenue'].sum()
axes[1, 0].pie(revenue_by_payment, labels=revenue_by_payment.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
axes[1, 0].set_title('Revenue Share by Payment Method', fontsize=14, fontweight='bold')

# 4. Sales Trend Over Time
sales_trend = df_clean.groupby('Sale_Date')['Total_Revenue'].sum().reset_index()
sns.lineplot(x='Sale_Date', y='Total_Revenue', data=sales_trend, ax=axes[1, 1], marker='o', color='purple')
axes[1, 1].set_title('Daily Sales Revenue Trend', fontsize=14, fontweight='bold')
axes[1, 1].set_xlabel('Date')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
